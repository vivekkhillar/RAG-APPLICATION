```python
from fastapi import FastAPI
import chromadb

def main():
    app = FastAPI()

    client = chromadb.HttpClient(
        host="localhost",
        port=8001   # use your actual ChromaDB port
    )

    @app.get("/chroma/collections")
    def get_collections():

        collections = client.list_collections()

        result = []

        for collection in collections:
            result.append({
                "name": collection.name,
                "count": collection.count()
            })

        return {
            "collections": result
        }


if __name__ == "__main":
    app = main()
    print(app)
```
